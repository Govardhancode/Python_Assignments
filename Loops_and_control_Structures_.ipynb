Problem 3.1: Reverse a Number (Medium)

```python
n = int(input("Enter a number: "))

num = abs(n)
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if n < 0:
    reverse = -reverse

print("Reversed number =", reverse)
```

Problem 3.4: Palindrome Checker (Medium)

```python
num = int(input("Enter a number: "))
original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print(original, "is a palindrome number")
else:
    print(original, "is not a palindrome number")
```

Problem 3.5: Armstrong Number (Hard)

```python
num = int(input("Enter a number: "))
original = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total += digit ** digits
    num = num // 10

if total == original:
    print(original, "is an Armstrong number")
else:
    print(original, "is not an Armstrong number")
```

Problem 6.4: LCM Calculator

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

x, y = abs(a), abs(b)

while y != 0:
    remainder = x % y
    x = y
    y = remainder

gcd = x

if a == 0 or b == 0:
    lcm = 0
else:
    lcm = abs(a * b) // gcd

print("LCM:", lcm)
```

Problem 6.5: Number to Words (0–999)

```python
ones = [
    "Zero", "One", "Two", "Three", "Four",
    "Five", "Six", "Seven", "Eight", "Nine"
]

teens = [
    "Ten", "Eleven", "Twelve", "Thirteen", "Fourteen",
    "Fifteen", "Sixteen", "Seventeen", "Eighteen", "Nineteen"
]

tens = [
    "", "", "Twenty", "Thirty", "Forty",
    "Fifty", "Sixty", "Seventy", "Eighty", "Ninety"
]

num = int(input("Enter a number (0-999): "))

if num < 0 or num > 999:
    print("Please enter a number between 0 and 999.")
elif num == 0:
    print("Zero")
else:
    words = []

    if num >= 100:
        words.append(ones[num // 100])
        words.append("Hundred")
        num %= 100

    if num >= 20:
        words.append(tens[num // 10])
        num %= 10

    elif num >= 10:
        words.append(teens[num - 10])
        num = 0

    if num > 0:
        words.append(ones[num])

    print(" ".join(words))
```

Problem 6.6: Strong Number Checker

```python
num = int(input("Enter a number: "))
original = num
total = 0

while num > 0:
    digit = num % 10
    fact = 1

    for i in range(1, digit + 1):
        fact *= i

    total += fact
    num //= 10

if original >= 0 and total == original:
    print(original, "is a Strong number")
else:
    print(original, "is not a Strong number")
```

Problem 7.1: Collatz Conjecture

```python
n = int(input("Enter a positive integer: "))
steps = 0

if n <= 0:
    print("Please enter a positive integer.")
else:
    print("Sequence:", end=" ")

    while n != 1:
        print(n, end=" -> ")

        if n % 2 == 0:
            n //= 2
        else:
            n = 3 * n + 1

        steps += 1

    print(1)
    print("Number of steps:", steps)
```

Problem 7.2: Caesar Cipher

```python
message = input("Enter a message: ")
shift = int(input("Enter shift value: ")) % 26
choice = input("Encrypt or decrypt (E/D): ").upper()

if choice == "D":
    shift = -shift
elif choice != "E":
    print("Invalid choice!")
    exit()

result = ""

for ch in message:
    if ch.isupper():
        result += chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
    elif ch.islower():
        result += chr((ord(ch) - ord('a') + shift) % 26 + ord('a'))
    else:
        result += ch

```

Problem 7.3: ATM Machine Simulator

```python
balance = 10000
pin = "1234"
history = []
attempts = 0

while attempts < 3:
    entered_pin = input("Enter PIN: ")

    if entered_pin == pin:
        break

    attempts += 1
    print("Incorrect PIN!")

if attempts == 3:
    print("Card blocked. Too many incorrect attempts.")
else:
    while True:
        print("\n1. Check Balance")
        print("2. Withdraw Money")
        print("3. Deposit Money")
        print("4. Change PIN")
        print("5. Transaction History")
        print("6. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            print("Balance:", balance)

        elif choice == "2":
            amount = float(input("Enter withdrawal amount: "))

            if amount <= 0:
                print("Invalid amount!")
            elif amount > balance:
                print("Insufficient balance!")
            else:
                balance -= amount
                history.append(f"Withdrew: {amount:.2f}")
                print("Withdrawal successful.")
                print("Balance:", balance)

        elif choice == "3":
            amount = float(input("Enter deposit amount: "))

            if amount <= 0:
                print("Invalid amount!")
            else:
                balance += amount
                history.append(f"Deposited: {amount:.2f}")
                print("Deposit successful.")
                print("Balance:", balance)

        elif choice == "4":
            old_pin = input("Enter current PIN: ")

            if old_pin == pin:
                new_pin = input("Enter new 4-digit PIN: ")

                if new_pin.isdigit() and len(new_pin) == 4:
                    pin = new_pin
                    history.append("PIN changed")
                    print("PIN changed successfully.")
                else:
                    print("PIN must contain 4 digits.")
            else:
                print("Incorrect current PIN.")

        elif choice == "5":
            print("Last 5 transactions:")
            for item in history[-5:]:
                print(item)

        elif choice == "6":
            print("Thank you for using the ATM.")
            break

        else:
            print("Invalid choice!")
```

Problem 11.3: Spiral Matrix

```python
n = int(input("Enter matrix size: "))
matrix = [[0] * n for _ in range(n)]

top, bottom = 0, n - 1
left, right = 0, n - 1
num = 1

while top <= bottom and left <= right:
    for j in range(left, right + 1):
        matrix[top][j] = num
        num += 1
    top += 1

    for i in range(top, bottom + 1):
        matrix[i][right] = num
        num += 1
    right -= 1

    if top <= bottom:
        for j in range(right, left - 1, -1):
            matrix[bottom][j] = num
            num += 1
        bottom -= 1

    if left <= right:
        for i in range(bottom, top - 1, -1):
            matrix[i][left] = num
            num += 1
```

Problem 11.5: Twin Primes

```python
n = int(input("Enter limit: "))

for num in range(2, n - 1):
    prime1 = True
    prime2 = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            prime1 = False
            break

    second = num + 2

    for i in range(2, int(second ** 0.5) + 1):
        if second % i == 0:
            prime2 = False
            break

    if prime1 and prime2 and second <= n:
        print(num, second)
```

Problem 11.8: Sieve of Eratosthenes

```python
n = int(input("Enter limit: "))
is_prime = [True] * (n + 1)

if n >= 0:
    is_prime[0] = False
if n >= 1:
    is_prime[1] = False

for i in range(2, int(n ** 0.5) + 1):
    if is_prime[i]:
        for j in range(i * i, n + 1, i):
            is_prime[j] = False

print("Prime numbers:")
for i in range(2, n + 1):
    if is_prime[i]:
        print(i, end=" ")
print()
```

Problem 11.10: Sudoku Validator

```python
grid = []

print("Enter 9 rows, with 9 digits separated by spaces:")

for i in range(9):
    row = list(map(int, input().split()))
    grid.append(row)

valid = all(
    len(row) == 9 and set(row) == set(range(1, 10))
    for row in grid
)

if valid:
    for j in range(9):
        column = [grid[i][j] for i in range(9)]
        if set(column) != set(range(1, 10)):
            valid = False
            break

if valid:
    for i in range(0, 9, 3):
        for j in range(0, 9, 3):
            box = []
            for r in range(i, i + 3):
                for c in range(j, j + 3):
                    box.append(grid[r][c])

            if set(box) != set(range(1, 10)):
                valid = False
                break
        if not valid:
            break

if valid:
    print("Valid Sudoku")
else:
    print("Invalid Sudoku")
```
