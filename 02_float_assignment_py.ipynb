```python
# # Question 1: Calculate the average of 3.14, 2.718, 1.618, 0.577

a = 3.14
b = 2.718
c = 1.618
d = 0.577
average = (a + b + c + d) / 4
print("Average =", average)
```

```python
# Question 2: Convert 98.6 Fahrenheit to Celsius

fahrenheit = 98.6
celsius = (fahrenheit - 32) * 5 / 9
print("Celsius =", celsius)
```

```python
# Question 3: Calculate compound interest on $1000 at 5.5% for 3 years

principal = 1000
rate = 5.5
time = 3

amount = principal * (1 + rate / 100) ** time
compound_interest = amount - principal

print("Compound Interest =", compound_interest)
```

```python
# Question 4: Find hypotenuse with sides 3.5 and 4.2

import math    #important

a = 3.5
b = 4.2

hypotenuse = math.sqrt(a ** 2 + b ** 2)

print("Hypotenuse =", hypotenuse)
```

```python
# Question 5: Calculate volume of sphere with radius 7.8


radius = 7.8

volume = (4 / 3) * math.pi * radius ** 3

print("Volume =", volume)
```

```python
# Question 6: Round 3.14159 to 3 decimal places

number = 3.14159

result = round(number, 3)

print("Rounded number =", result)
```

```python
# Question 7: Calculate percentage: 45 out of 67

obtained = 45
total = 67

percentage = (obtained / total) * 100

print("Percentage =", percentage)
```

```python
# Question 8: Find square root of 23.456

number = 23.456

square_root = math.sqrt(number)

print("Square root =", square_root)
```

```python
# Question 9: Calculate simple interest

principal = 2500
rate = 6.5
time = 2.5

simple_interest = (principal * rate * time) / 100

print("Simple Interest =", simple_interest)
```

```python
# Question 10: Convert 45.7 degrees to radians

degree = 45.7

radian = math.radians(degree)

print("Radians =", radian)
```
