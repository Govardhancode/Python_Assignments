```python
# Question 10: Remove all vowels from "Computer Science"

text = "Computer Science"
vowels = "aeiouAEIOU"

result = ""

for char in text:
    if char not in vowels:
        result += char

print(result)
```

```python
# Question 11: Find the most frequent character in "mississippi"

text = "mississippi"

result = max(text, key=text.count)

print(result)
```

```python
# Question 14: Count consonants in "Hello World"

text = "Hello World"
vowels = "aeiouAEIOU"

count = 0

for char in text:
    if char.isalpha() and char not in vowels:
        count += 1

print(count)
```

```python
# Question 15: Find the longest word in "Python is a programming language"

text = "Python is a programming language"

words = text.split()

print(max(words, key=len))
```

```python
# Question 16: Remove all punctuation from "Hello, World! How are you?"

import string

text = "Hello, World! How are you?"

result = ""

for char in text:
    if char not in string.punctuation:
        result += char

print(result)
```

```python
# Question 23: Convert "hello world" to alternating case

text = "hello world"

result = ""

for i in range(len(text)):
    if i % 2 == 0:
        result += text[i].lower()
    else:
        result += text[i].upper()

print(result)
```

```python
# Question 24: Find all positions of 'a' in "banana"

text = "banana"

positions = []

for i in range(len(text)):
    if text[i] == "a":
        positions.append(i)

print(positions)
```

```python
# Question 29: Count words that start with 'p' in "Python programming is powerful"

text = "Python programming is powerful"

count = 0

for word in text.split():
    if word.lower().startswith("p"):
        count += 1

print(count)
```

```python
# Question 34: Find common characters between "hello" and "world"

text1 = "hello"
text2 = "world"

common = set(text1) & set(text2)

print(common)
```

```python
# Question 35: Check if string is a valid phone number: "+1-555-123-4567"

phone = "+1-555-123-4567"

if phone.startswith("+") and phone.replace("+", "").replace("-", "").isdigit():
    print(True)
else:
    print(False)
```

```python
# Question 36: Extract numbers from "abc123def456ghi789"

text = "abc123def456ghi789"

numbers = ""

for char in text:
    if char.isdigit():
        numbers += char

print(numbers)
```

```python
# Question 37: Convert "snake_case" to "camelCase"

text = "snake_case"

words = text.split("_")

result = words[0] + words[1].title()

print(result)
```

```python
# Question 38: Check palindrome ignoring case and spaces

text = "A man a plan a canal Panama"

text = text.replace(" ", "").lower()

print(text == text[::-1])
```

```python
# Question 39: Find the most common word in
# "the quick brown fox jumps over the lazy dog"

text = "the quick brown fox jumps over the lazy dog"

words = text.split()

print(max(words, key=words.count))
```

```python
# Question 40: Generate acronym from
# "National Aeronautics and Space Administration"

text = "National Aeronautics and Space Administration"

words = text.split()

acronym = ""

for word in words:
    acronym += word[0].upper()

print(acronym)
```
