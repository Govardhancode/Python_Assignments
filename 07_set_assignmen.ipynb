```python
print("Question 1: Create a set of first 10 natural numbers")
num  = range(1,11)
set_num = set(num)
print(f"First 10 natural number i {set_num}")
```

```python
print("\nQuestion 2: Add element 11 to set {1, 2, 3, 4, 5}")
numbers =  {1, 2, 3, 4, 5}
numbers.add(11)
print(numbers)
```

```python
print("\nQuestion 3: Remove element 3 from set {1, 2, 3, 4, 5, 6}")
numbers = {1, 2, 3, 4, 5, 6}
numbers.remove(3)
print(numbers)
```

```python
print("\nQuestion 4: Find the intersection of {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}")
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print(f"Intersection of two sets is{set1.intersection(set2)}")
```

```python
print("\nQuestion 5: Find the difference between {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}")
set1 = {1, 2, 3, 4, 5}
set2 = {4,5,6,7,8}
print(f" Minus of Two sets is{set1 - set2}")
```

```python
print("\nQuestion 6: Check if 5 is in set {1, 2, 3, 4, 6, 7, 8}")
set = {1, 2, 3, 4, 6, 7, 8}

if 5 in set:
    print("5 is available in set")
else:
    print("5 is not available in set")
```

```python
print("\nQuestion 7: Find the length of set {'a', 'b', 'c', 'd', 'e'}")
set = {'a', 'b', 'c', 'd', 'e'}
print(f"length of set is {len(set)}")
```

```python
print("\nQuestion 8: Create a set of vowels from string 'hello world'")
text = "hello world"
vowels = {'a', 'e', 'i', 'o', 'u'}
vowel_set = {ch for ch in text if ch in vowels}

print(vowel_set)
```

```python
print("\nQuestion 9: Remove duplicates from list [1, 2, 2, 3, 4, 4, 5, 6, 6, 7] using set")
list = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
set_list = set(list)
print(set_list)
```
