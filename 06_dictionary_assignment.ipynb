```python
# Question 44: Find sum of each nested list

data = {
    "matrix": [[1, 2], [3, 4]],
    "vector": [5, 6, 7]
}

matrix_sums = []

for row in data["matrix"]:
    matrix_sums.append(sum(row))

vector_sum = sum(data["vector"])

print("Matrix sums:", matrix_sums)
print("Vector sum:", vector_sum)
```

```python
# Question 49: Create a dictionary with nested sets
# Sets cannot contain other sets directly,
# so we use lists containing sets.

data = {
    "groups": [
        {1, 2, 3},
        {4, 5, 6}
    ],
    "categories": [
        {"a", "b"},
        {"c", "d"}
    ]
}

print(data)
```

```python
# Question 50: Find union of all nested sets

data = {
    "groups": [
        {1, 2, 3},
        {4, 5, 6}
    ],
    "categories": [
        {"a", "b"},
        {"c", "d"}
    ]
}

groups_union = set()
categories_union = set()

for item in data["groups"]:
    groups_union = groups_union.union(item)

for item in data["categories"]:
    categories_union = categories_union.union(item)

print("Groups union:", groups_union)
print("Categories union:", categories_union)
```
